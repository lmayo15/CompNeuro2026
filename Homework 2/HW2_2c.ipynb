# Find numerically, to three digits, the voltage at which the slope vanishes for $[Mg^{2+}] = 1 mM$, plot $I_N(V)$ over $V ∈ [−90, 0]$ mV for $[Mg^{2+}] = 0, 0.1, 1 mM$, and explain what the sign of the slope below that voltage means for the synapse.

```python
import numpy as np
import matplotlib.pyplot as plt

V = np.linspace(-90, 0, 100)    # mV
V_0 = 16.13     # mV
E_N = 0     # mV

Mg = [0, 0.1, 1]    #mM

# Find values and create plot
for c in Mg:   
    B = 1 / (1 + (c / 3.57) * np.exp(-V / V_0))
    InV = B * (V - E_N) 
    
    plt.plot(V, InV, label = f"[Mg2+] = {c} mM")

# Labeling
plt.xlabel("V (mV)")
plt.ylabel("$I_N\  /\  \overline{g}$")
plt.title("NMDA Current for Varying Magnesium Concentrations")
plt.legend()
plt.show()
```
