# Plotting $c$ against $a$ over $ 0 < a < 1$ at $D=1$

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.linspace(0.001, 0.999, 1000)
c = (1)((1-2*a)/np.sqrt(a*(1-a)))

cross = (1-1/np.sqrt(2))/2

# plot design
plt.figure(figsize = 7, 4)
plt.plot(a, c, label = r"$c=(1-2a)/\sqrt{a(1-a)}")
plt.axhline(2, linestyle = "--", label = "Passive apparent speed = 2")
plt.axhline(cross, linestyle = ":", label = "fr:Crossing: $a={cross:.5f}$")
plt.axhline(0, linewidth = 0.8)
plt.xlim(0, 1)
plt.ylim(-14, 14)
plt.xlabel(r"Threshold $a$")
plt.ylabel(r"Active front speed $c$ ($D = 1$)")
plt.title("Active Front Speed c vs Threshold a")
plt.legend()
plt.show()
```
